```python
%%writefile hf_inference.py
#!/usr/bin/env python3
"""
Hugging Face inference helper: loads a model+tokenizer, runs inference,
with logging, batching, and optional GPU.
"""

import argparse
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import logging

# Simple logger (replacement for utils.logging_utils)
def get_logger(name="hf_inference"):
    logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
    return logging.getLogger(name)

logger = get_logger("hf_inference")

def run_inference(model_name: str, texts: list, max_new_tokens: int = 64, batch_size: int = 1, device: str = None):
    """Run inference on Hugging Face model"""
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    logger.info(f"Loading model {model_name} on {device}")

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForCausalLM.from_pretrained(model_name).to(device)

    outputs = []
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i+batch_size]
        inputs = tokenizer(batch, return_tensors="pt", padding=True).to(device)
        with torch.no_grad():
            gen_tokens = model.generate(**inputs, max_new_tokens=max_new_tokens)
        decoded = tokenizer.batch_decode(gen_tokens, skip_special_tokens=True)
        outputs.extend(decoded)

    return outputs

if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--model", type=str, required=True, help="Model name (e.g. google/gemma-2-2b-it)")
    parser.add_argument("--texts", type=str, nargs="+", required=True, help="Input texts")
    parser.add_argument("--max-new-tokens", type=int, default=64, help="Max new tokens to generate")
    parser.add_argument("--batch", type=int, default=1, help="Batch size")
    args = parser.parse_args()

    results = run_inference(args.model, args.texts, args.max_new_tokens, args.batch)
    for r in results:
        print("="*40)
        print(r)

```
